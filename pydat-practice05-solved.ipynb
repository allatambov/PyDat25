Для более универсального выбора строк и столбцов на датафреймах `pandas` определены методы `.loc` и `.iloc`. Название метода `.loc` происходит от *location*, а метода `.iloc` – от *index location*, поэтому:

* `.loc` выбирает элементы по индексам строк (целые числа или текст) и названиям столбцов (текст);
* `.iloc` выбирает элементы по индексам строк (целые числа) и индексам столбцов (целые числа).

Так как эти методы являются дополнением к операторам `[]` для фильтрации, при их вызове тоже используются квадратные скобки, а не круглые. На первом месте в квадратных скобках указываются идентификаторы строк, на втором – идентификаторы столбцов, например:

    news.loc[0, "author"]
    news.iloc[0, 2]
    
> Используя сначала метод `.loc`, а затем `.iloc`, выведите на экран: сложность новости с индексом 50,  заголовок новости с индексом 1727, заголовок новости с индексом 1744. 

```python
# на первом месте – индекс строки (номер)
# на втором месте – название столбца

print(news.loc[50, "diffc"])
print(news.loc[1727, "title"])
print(news.loc[1744, "title"])
```

```python
# на первом месте – индекс строки (номер)
# на втором месте – индекс столбца (номер)

print(news.iloc[50, 4])
print(news.iloc[1727, 1])
print(news.iloc[1744, 1])
```

> Подумайте, как запросить элементы сразу по нескольким индексам/названиям. Используя `.loc`, выведите значение сложности и автора для новости с индексом 20. Используя `.iloc`, выведите автора и дату новостей с индексами 10, 20, 30.

```python
# если несколько названий – названия внутри списка
# строка с номером 20, столбцы diffc и author

news.loc[20, ["diffc", "author"]]
```

**Комментарий.** Pandas выбрал строку с номером 20, из нее – нужный фрагмент со сложностью и именем автора:

|    |   diffc | author       |
|---:|--------:|:-------------|
| 20 |     2.4 | Андрей Фокин |

А затем транспонировал результат – записал строку в виде столбца. Выше в выдаче мы видим последовательность `pandas Series`, столбец с двумя записями: индекс `diffc` и значение 2.4, индекс `author` и значение Андрей Фокин. В названии `Name` сохранен номер строки, которую мы выбрали изначально. А вот с типом столбца все любопытно. В `dtype` написано `object`, Python намекает, что в столбце сохранены текстовые значения. В теории так и должно быть – если в столбце есть и текст, и числа, текстовый тип `object`, как более сильный, должен вытеснить все остальные. Однако в действительности этого не происходит. Хотя Python и написал `object`, значения в последовательности сохранили свои типы:

```python
res = news.loc[20, ["diffc", "author"]]

# сложность осталась числом типа float

print(type(res["diffc"]))
print(type(res["author"]))
```

Это сделано для удобства – ценность `.loc` и `.iloc` была бы сомнительной, если бы после фильтрации с их помощью пришлось заново «восстанавливать» все типы, чтобы полноценно работать с выбранными фрагментами таблицы.

```python
# строки с номерами 10, 20, 30
# столбцы с номерами 2 и 3

news.iloc[[10, 20, 30], [2, 3]]
```

Как быть, если нужные нам индексы или названия следуют друг за другом? Воспользоваться срезами (*slices*), как на списках или массивах!

> Выведите на экран заголовки и авторов новостей с индексами от 610 до 614 включительно. Используйте сначала `.loc`, а затем – `.iloc`.

```python
# loc - правая граница среза включается
# строки с номерами от 610 до 614 включительно,
# столбцы от title до author включительно

news.loc[610:614, "title":"author"]
```

```python
# iloc - правая граница среза НЕ включается
# строки с номерами от 610 до 615, исключая 615
# столбцы с номерами от 1 до 3, исключая 3

news.iloc[610:615, 1:3]
```

**Комментарий.** Так как метод `.iloc` работает исключительно с целочисленными индексами, он «наследует» поведение числовых срезов и функции `range()` в базовом Python – исключает правую границу диапазона, который мы выбираем. Метод `.loc` работает с текстовыми значениями в том числе, поэтому в нем срезы работают несколько иначе.

> Как выбрать все строки и один столбец, используя срезы? Или одну строку, но все столбцы? Выберите столбец `diffc`, а затем – строку с индексом 3.

```python
# : – все строки от 0 до последней
news.loc[:, "diffc"]
```

```python
# : – все столбцы от 0 до последнего

news.loc[3, :]
```

## Часть 2: простое объединение датафреймов через `concat()`

Как мы уже видели, одна новость может относиться к нескольким рубрикам сразу:

```python
print(news["rubrics"].unique())
```

> Используя метод `.str.split()`, получите на основе столбца с рубриками новостей `rubrics` новые столбцы, где названия рубрик указаны по-отдельности. Сохраните новые столбцы в датафрейм `rubs`, присвойте им новые названия, записав их в атрибут датафрейма `.columns`.

```python
rubs = news["rubrics"].str.split(", ", expand = True)
rubs.columns = ["rub1", "rub2", "rub3", "rub4"]
rubs
```

> Используя функцию `concat()` из `pandas`, объедините датафреймы `news` и `rubs` в один итоговый датафрейм `final`. Выгрузите его в файл Excel.

```python
# склеить по столбцам (axis = 1),
# а не по строкам (axis = 0 по умолчанию)

final = pd.concat([news, rubs], axis = 1)
final.head()
```

```python
final.to_excel("news.xlsx")
```

Это рациональнее, чем вручную заменять все `\\` на корректно воспринимаемые `/` :)
