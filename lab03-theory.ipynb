### Множества в Python

Пусть определены два списка:

```python
one = [1, 2, 6, 7, 8, 10]
two = [4, 3, 8, 1, 12, 5, 7]
```

Наша задача – выбрать из списков `one` и `two` общие элементы. Если мы будем решать эту задачу только с помощью списков и циклов, нам придется перебрать все элементы списка `one`, затем – все элементы списка `two`, и сравнить все возможные пары значений. Для этого придется писать цикл в цикле:

```python
common = []

for i in one:
    for j in two:
        if i == j:
            common.append(i)

# выводим отсортированный список            
print(sorted(common))
```

Это не очень удобно и с точки зрения работы кода не очень быстро. Гораздо удобнее превратить списки в множества (тип `set`) и выполнять операции, которые определены для множеств в математике: пересечение, объединение, разность и так далее. 

Переделаем списки `one` и `two` в множества `A` и `B`:

```python
A = set(one)
B = set(two)
```

**Важно.** Стоит иметь в виду, что во множествах (как в программировании, так и в математике) не может быть повторяющихся элементов. Поэтому, если в списке есть повторяющиеся значения, и они важны, превращать спискок в множество не стоит, так как можно потерять элементы. Однако в нашем примере все элементы в `one` и `two` встречаются по одному разу, потери нам не грозят.

```python
# а вот пример с потерями
# потеряли повторы 1, 2, 6

set([1, 2, 6, 6, 1, 2])
```

Вернемся к нашей задаче. Раз нам нужны общие элементы множеств, надо найти их **пересечение $A \cap B$**:

```python
# метод .intersection()
# заметим, что порядок множеств здесь не важен

print(A.intersection(B))
print(B.intersection(A))
```

Заметим, что порядок множеств здесь не важен:

Если на выходе нам нужен отсортированный список общих значений, можем применить функцию `sorted()`, она принимает на вход любую последовательность, а возвращает всегда список (тип `list`):

```python
AB = sorted(A.intersection(B))
print(AB)
```

Убедились, что решение задачи с использованием множеств выглядит более лаконично и приятно, чем вариант с вложенным циклом. Рассмотрим другие примеры операций на множествах:

**Объединение** множеств `A` и `B` – все элементы множества `A` и множества `B`, без повторений ($A \cup B$):

```python
# метод .union()
# заметим, что порядок множеств здесь не важен

print(A.union(B))
print(B.union(A))
```

**Разность** множеств `A` и `B` – все элементы множества `A`, которых нет в `B` ($A\text{\\}B$):

```python
# метод .difference()
# вот тут порядок важен, A/B не B/A

print(A.difference(B))
```

**Разность** множеств `B` и `A` – все элементы множества `B`, которых нет в `A` ($B\text{\\}A$):

```python
print(B.difference(A))
```

**Симметрическая разность** множеств `A` и `B` – все элементы `A` и `B`, которые не входят в пересечение этих множеств, то есть все элементы, которые есть только в `A` или только в `B` ($A \triangle B$):

```python
# метод .symmetric_difference()
# и тут порядок множеств снова не важен 
# разность симметрическая – симметрия

print(A.symmetric_difference(B))
print(B.symmetric_difference(A))
```

Также на множествах определены особые методы, которые облегчают сравнение наборов элементов. Для удобства создадим новые множества `C`, `D` и `E`, где множество `C` полностью входит в множество `D`, то есть является его подмножеством.

```python
C = {5, 8, 9}
D = {20, 5, 8, 9, 10, 30}
E = {10, 2, 7}
```

Так, можно проверить, являются ли множества непересекающимися (не имеющими общих элементов):

```python
# joint - пересекающиеся, disjoint - непересекающиеся

print(C.isdisjoint(D))
print(C.isdisjoint(E))
```

Или проверить, является ли множество `C` подмножеством `D` (все элементы `C` входят в `D`):

```python
print(C.issubset(D))
```

И наоборот, включает ли множество `D` множество `C` или `E`:

```python
print(D.issuperset(C))
print(D.issuperset(E))
```

В заключение заметим, что множества могут состоять не только из чисел. Допустим, пользователь ввел с клавиатуры перечни, которые сохранились как строки:

```python
products01 = "сыр, масло, молоко, колбаса, йогурт"
products02 = "йогурт, йогурт, молоко, яблоко, сыр"
```

Если нам нужны общие элементы этих перечней без повторений, можем применить метод `.split()`, разбить строки по запятой с пробелом, а затем переделать полученные списки в множества и найти их пересечение:

```python
s1 = set(products01.split(", "))
s2 = set(products02.split(", "))
print(s1, s2)
```

```python
joint = s1.intersection(s2)
print(sorted(joint))
```
