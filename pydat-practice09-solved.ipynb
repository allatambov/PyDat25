## Практикум 9. Датафреймы pandas: агрегирование и введение в мультииндексы

В файлах `Dialogue.csv`-`Movies.csv` (источник – Kaggle) сохранена информация по фильмам о Гарри Поттере:

* `Dialogue`: реплика героя в фильме;
* `Dialogue ID`: ID реплики;
* `Chapter ID`: ID сцены, в которой произносится реплика;
* `Chapter Name`: название сцены, в которой произносится реплика;
* `Place ID`: ID локации, где была произнесена реплика;
* `Place Name`: название локации, где была произнесена реплика;
* `Place Category`: тип локации где была произнесена реплика;
* `Character ID`: ID героя;
* `Character Name`: имя героя;
* `Species`, `Gender`, `House`: вид, пол, факультет/школа героя;
* `Patronius`: патронус героя;
* `Wand (Wood)` и `Wand (Core)`: материал и сердцевина палочки героя;
* `Movie Title`: название фильма;
* `Movie ID`: ID фильма;
* `Release Year`, `Runtime`: год выпуска и продолжительность фильма в минутах;
* `Budget` и `Box Office`: бюджет и кассовый сбор фильма.

```python
import pandas as pd

dialog = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyDat25/refs/heads/main/Dialogue.csv")
chapters = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyDat25/refs/heads/main/Chapters.csv")
heroes = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyDat25/refs/heads/main/Characters.csv")
places = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyDat25/refs/heads/main/Places.csv")
movies = pd.read_csv("https://raw.githubusercontent.com/allatambov/PyDat25/refs/heads/main/Movies.csv")
```

```python
display(dialog.head(3))
display(heroes.head(3))
display(chapters.head(3))
display(places.head(3))
display(movies.head(3))
```

### Задача 1

Объедините датафреймы `dialog` и `heroes` таким образом, чтобы к каждой реплике героя в `dialog` были добавлены характеристики героя из `heroes`.

Выполните разные варианты объединений (`left`, `right` и `inner`) и сравните результаты. Сохраните результат правильного объединения в датафрейм `one`.

```python
# правильный - сохраняется порядок строк как в dialog

one = dialog.merge(heroes, how = "left", on = "Character ID")
one
```

```python
# глобально объединение тоже, но теперь главный
# датафрейм heroes, поэтому строки упорядочены по героям,
# идут в том порядке, как в heroes

dialog.merge(heroes, how = "right", on = "Character ID")
```

```python
# ключевой датафрейм dialog, герои упорядочиваются 
# в том порядке, в каком они встречаются в dialog
# без повторений (сначала все реплики Дамблдора, 
# затем – все реплики МакГонагалл и т д)

dialog.merge(heroes, how = "inner", on = "Character ID")
```

### Задача 2

Добавьте признак `Length` – длину реплики, измеренную в символах. 

a. Определите, представители какого факультета/школы (`House`), в среднем, произносят самые длинные реплики.

b. Определите топ-5 героев – представителей факультета/школы из предыдущего пункта, которым принадлежат самые длинные реплики.

```python
# добавляем столбец

one["Length"] = one["Dialogue"].apply(lambda x: len(x))
one["Length"].head()
```

```python
# внезапно, но это не Гриффиндор с речами Дамблдора,
# а Дурмстранг (Крам заговорил, не иначе)

print(one.groupby("House")["Length"].mean())
```

```python
# определяем название (индекс) максимального значения 

top = one.groupby("House")["Length"].mean().idxmax()
print(top)
```

```python
# выбираем строки, соответствующие ему, и сортируем
# топ-5 не понадобился, тут всего 5 героев, 
# и самая длинная реплика принадлежит Гриндевальду

one[one["House"] == top].sort_values("Length", ascending = False)
```

### Задача 3

Определите, кто чаще называет Того-Кого-Нельзя-Называть Волан-де-Мортом: Гарри или Дамблдор. Другими словами, сравните количество реплик, содержащих слово `Voldemort`, принадлежащих героям `Harry Potter` и `Albus Dumbledore`.

```python
# вариант 1

chosen_one = one[one["Character Name"] == "Harry Potter"]
great_man_dumbledore = one[one["Character Name"] == "Albus Dumbledore"]

# Гарри

print(chosen_one["Dialogue"].str.contains("Voldemort", 
                                          case = False).sum())
print(great_man_dumbledore["Dialogue"].str.contains("Voldemort", 
                                          case = False).sum())
```

```python
# вариант 2

res = one[one["Dialogue"].str.contains("Voldemort", case = False) & 
          one["Character Name"].isin(["Harry Potter", "Albus Dumbledore"])]

# Гарри

res.groupby("Character Name")["Dialogue"].count()
```

### Задача 4

Объедините датафреймы таким образом, чтобы к каждой реплике была добавлена информация о том, в какой локации она произнесена и в каком фильме.

a. Определите, в каком фильме (`Movie Title`) было задействовано больше всего локаций (`Place Name`). 

b. Сформируйте сводную таблицу, в которой для каждого фильма посчитано, сколько локаций каждого типа (`Place Category`) там задействовано. Используйте функцию `pd.crosstab()`.

```python
full = one.merge(chapters, 
                 on = "Chapter ID").merge(places, 
                                          on = "Place ID").merge(movies, 
                                                                 on = "Movie ID")
```

```python
# пункт a
# группируем по названию фильма,
# считаем число заполненных ячеек в Place Name,
# находим индекс максимального значения

full.groupby("Movie Title")["Place Name"].count().idxmax()
```

```python
# пункт a: дополнение
# если хочется учесть только уникальные названия локаций,
# чтобы оценить разнообразие локаций

full.groupby("Movie Title")["Place Name"].unique().apply(lambda x: len(x))
```

```python
# пункт b

pd.crosstab(full["Movie Title"], full["Place Category"])
```

Выполните пункт b из предыдущей задачи, используя только группировку. Изучите полученный объект и подумайте, как из него извлекать элементы.

### Мультииндексы

Это объект типа `pandas Series`, обычный столбец со значениями, однако индексы здесь более сложные – они состоят из двух уровней, название фильма `Movie Title` и тип локации `Place Category`. Такие индексы называются *мультииндексами*. Разберем столбец на части:

```python
# индексы – тип MultiIndex, внутри пары значений,
# первый элемент в паре – фильм: второй – тип локации

print(res.index)
```

```python
# значения – просто числовой массив
# количество локаций

print(res.values)
```

Как выбирать элементы по индексам, если индексы теперь записаны парами? Если нужно значение из конкретной  ячейки – парами и выбирать:

```python
# название фильма, затем – тип локации

print(res[("Harry Potter and the Chamber of Secrets", "Hogwarts")])
print(res[("Harry Potter and the Philosopher's Stone", "Diagon Alley")])
```

Если нужно несколько значений, например, все данные по фильму `Harry Potter and the Chamber of Secrets`, можно воспользоваться `.loc` и срезами. При работе с датафреймом в `.loc` мы указывали название/номер строки и название столбца, а здесь, раз мы имеем дело с одним столбцом, на первом месте мы можем указать индекс первого уровня (навзание фильма), а на втором – индекс второго уровня (тип локации):

```python
# первый уровень – название фильма
# второй уровень – все типы локаций

res.loc["Harry Potter and the Chamber of Secrets", :]
```

```python
# первый уровень – все фильмы
# второй уровень – один локации

res.loc[:, "Hogwarts"]
```

Альтернативный вариант – использование метода `.xs()`:

```python
# выбери данные, где на первом уровне (level = 0)
# указано название фильма Harry Potter and the Chamber of Secrets,
# вместо номера уровня можно вписать его название

print(res.xs("Harry Potter and the Chamber of Secrets", level = 0))
print(res.xs("Harry Potter and the Chamber of Secrets", level = "Movie Title"))
```

```python
# выбери данные, где на втором уровне (level = 1)
# указано Hogwarts
# вместо номера уровня можно вписать его название

print(res.xs("Hogwarts", level = 1))
print(res.xs("Hogwarts", level = "Place Category"))
```

### Длинный и широкий формат представления данных

Результат группировки `res` выше – это пример длинного формата представления сгруппированных данных (*long format*). Здесь данные по группам идут друг за другом в один столбец. Сначала идет перечень типов локаций в первом фильме, потом – во втором, и так далее. 

Таблица сопряженности, полученная в задаче 4 – пример широкого формата представления сгруппированных данных (*wide format*). Здесь данные по разным группам записаны в отдельных столбцах (один столбец соответствует одному типу локации):

```python
pd.crosstab(full["Movie Title"], full["Place Category"])
```

Оба эти формата бывают полезны для разных типов задач, при этом важно помнить, что из одного формата можно всегда получить другой с помощью метода `.unstack()`:

```python
# из длинного в широкий, long to wide,
# образуются пропуски, потому что не все локации есть во всех фильмах

res.unstack()
```

```python
# из широкого в длинный, wide to long

pd.crosstab(full["Movie Title"], full["Place Category"]).unstack()
```
