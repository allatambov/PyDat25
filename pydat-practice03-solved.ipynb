### Задача 0

Выведите информацию о размерности датафрейма и типах столбцов в датафрейме. Выведите описательные статистики для всех столбцов датафрейма.

Подсказка: описательные статистики все сразу – метод `.describe()` и этот же метод с аргументом `include = "object"`.

> Здесь `count` – число заполненных ячеек, `mean` – среднее арифметическое, `std` – стандартное отклонение, `min` и `max` – минимум и максимум, `50%` – медиана, `25%` – 25-ый процентиль, он же нижний квартиль, он же первый квартиль, он же квантиль уровня 0.25, `75%` – 75-ый процентиль, он же верхний квартиль, он же третий квартиль, он же квантиль уровня 0.75.

```python
# включаем в выдачу описание только текстовых столбцов
# типа object

stars.describe(include = "object")
```

> Здесь `count` – число заполненных ячеек, `unique` – число уникальных значений (категорий), `top` – мода, самое частое значение, `freq` – частота, соответствующая моде. Так, например, звезды по спектральному классу делятся на 7 категорий, самая частая из них – категория `M`, таких звезд в файле 111.

### Задача 1

Сохраните строки, соответствующие звездам с абсолютной звездной величиной (`Absolute magnitude`) ниже 0, в датафрейм `low`. Вычислите количество таких строк.

### Задача 2

Проверьте, есть ли в данных звезды, которые являются белыми карликами, чей цвет отличен от белого (значение `White`).

### Задача 3

Сохраните строки, соответствующие звездам спектрального класса `O` и `M` , в датафрейм `stars_om`. Вычислите среднюю температуру таких звезд и округлите ее до второго знака после запятой.

```python
# вариант 1

stars_om = stars[(stars["Spectral Class"] == "O") |
                 (stars["Spectral Class"] == "M")]
print(stars_om.shape)
```

```python
# вариант 2

stars_om = stars[stars["Spectral Class"].isin(["O", "M"])]
print(stars_om.shape)
```

```python
# считаем среднюю температуру
# второй вариант – если версия Python/Numpy старовата

print(stars_om["Temperature"].mean().round(2))
print(round(stars_om["Temperature"].mean(), 2))
```

### Задача 4

Сохраните строки, соответствующие звездам-сверхгигантам и звездам-супергигантам, в датафрейм `giants`. Вычислите минимальную светимость таких звезд.

```python
# вариант 1

giants = stars[(stars["Star type"] == 4) | (stars["Star type"] == 5)]
print(giants["Luminosity"].min())
```

```python
# вариант 2

giants = stars[stars["Star type"].isin([4, 5])]
print(giants["Luminosity"].min())
```

```python
# вариант 3

giants = stars[stars["Star type"] >= 4]
print(giants["Luminosity"].min())
```

### Задача 5

Сохраните строки, соответствующие звездам главной последовательности синего (`Blue`) или сине-белого цвета (`Blue-white` или `Blue-White`), в датафрейм `main_blue`.

Вычислите количество строк в полученном датафрейме. Выведите описательные статистики для столбца `Spectral Class`. Звезд какого спектрального класса больше всего среди отобранных наблюдений?

```python
# вариант 1: ключевой, обратите внимание
# на наличие дополнительных скобок

main_blue = stars[(stars["Star type"] == 3) &
      ((stars["Star color"] == "Blue") |
       (stars["Star color"] == "Blue-white") |
       (stars["Star color"] == "Blue-White"))]

print(main_blue.shape[0])
```

```python
# вариант 2: .isin() вместо ИЛИ

main_blue = stars[(stars["Star type"] == 3) &
                  (stars["Star color"].isin(["Blue", "Blue-white",
                                             "Blue-White"]))]

print(main_blue.shape[0])
```

```python
# вариант 3: если понравился str.contains(),
# пользуемся тем, что Blue-white и Blue-White
# отличаются только регистром, поэтому
# можем выключить учет регистра case = False


main_blue = stars[(stars["Star type"] == 3) &
     (stars["Star color"].str.contains("Blue") |
     stars["Star color"].str.contains("Blue-white", case = False))]
print(main_blue.shape[0])
```

```python
# больше всего звезд класса B

main_blue["Spectral Class"].describe()
```

### Задача 6

Сохраните строки, соответствующие звездам, чья температура не ниже медианы (50-ый процентиль), но не выше верхнего квартиля (75-ый процентиль), в датафрейм `last`.

```python
last = stars[(stars["Temperature"] >= stars["Temperature"].median()) &
     (stars["Temperature"] <= stars["Temperature"].quantile(0.75))]
print(last.shape)
```
