## Практикум 14. Корреляционная матрица и коэффициенты корреляции

### Корреляционная матрица

Импортируем `pandas` и загрузим данные из файла `Griliches.csv`:

```python
import pandas as pd
```

```python
df = pd.read_csv("Griliches.csv")
df.head()
```

Полное описание данных можно найти [здесь](https://vincentarelbundock.github.io/Rdatasets/doc/Ecdat/Griliches.html), это довольно древние данные, они используются в R как учебные данные для построения разных эконометрических моделей. Выберем классические показатели, которые часто изучают при рассмотрении факторов, от которых зависит заработная плата (здесь `lw` –  натуральный логарифм заработной платы):

* `age`: возраст;
* `expr`: опыт работы (стаж);
* `iq`:  коэффициент интеллекта IQ.

Так как все показатели числовые, можем построить для них **корреляционную матрицу**. По умолчанию это будет матрица с коэффициентами корреляции Пирсона:

```python
df[["age", "lw", "expr", "iq"]].corr()
```

Для удобства можем округлить все коэффициенты до сотых:

```python
df[["age", "lw", "expr", "iq"]].corr().round(2)
```

На главной диагонали матрицы стоят 1 – коэффициент Пирсона переменной самой с собой всегда равен 1. Почти все коэффициенты положительны: так, например, наблюдается умеренная положительная связь между возрастом и заработной платой (0.53), между возрастом и опытом работы (0.39), между заработной платой и уровнем IQ (0.35). Связь между заработной платой и опытом работы – слабая положительная (0.08), связь между возрастом и уровнем IQ – тоже (0.18). Слабая отрицательная связь слабой наблюдается между опытом работы и уровнем IQ (–0.17).

Для получения матрицы с коэффициентами корреляции Спирмена в метод `.corr()` нужно добавить аргумент `method = "spearman"`:

```python
df[["age", "lw", "expr", "iq"]].corr(method = "spearman").round(2)
```

Если в совместном распределении нет выбросов, значения коэффициента Пирсона и коэффициента Спирмена различаются несильно. Можем вычислить разницу между корреляционными матрицами (это два датафрейма одинаковой размерности) и убедиться в этом:

```python
df[["age", "lw", 
    "expr", "iq"]].corr() - df[["age", "lw", 
                                "expr", "iq"]].corr(method = "spearman")
```

**NB.** Если хочется оценить связь «всего со всем», то есть построить корреляционную матрицу для всех числовых переменных в датафрейме, в новых версиях `pandas` необходимо гарантировать, что мы применяем метод `.corr()` к датафрейму с соответствующими типами столбцов. Чтобы выбрать все числовые столбцы, можно воспользоваться методом `.select_dtypes()`:

```python
# выбираем столбцы типов int, float

df.select_dtypes([int, float])
```

```python
# применяем .corr()

df.select_dtypes([int, float]).corr()
```

### Коэффициенты корреляции по группам

Данные в `df` – кросс-секционные, в таблице хранятся наблюдения за разные годы. Иногда имеет смысл оценивать связи между переменными по каждому году отдельно. Сфокусируемся на паре показателей `age` и `lw` и вычислим коэффициент Пирсона между ними за каждый год (столбец `year`):

```python
# группируем по year,
# в каждой группе выбирем столбцы age, lw,
# применяем .corr()

df.groupby("year")[["age", "lw"]].corr()
```

Чисто технически, задача выполнена. Однако полученные результаты избыточны – метод `.corr()` возвращает корреляционную матрицу целиком, а нам достаточно одного коэффициента (1 на диагонали нас не интересует, а сам коэффициент дублируется). Чтобы понять, как извлечь один коэффициент, перепишем код выше, используя `.apply()` и lambda-функцию:

```python
# тот же итог, что и выше

df.groupby("year")[["age", "lw"]].apply(lambda x: x.corr().round(2))
```

Что является результатом применения написанной lambda-функции к одной группе в `.groupby()`? Маленький датафрейм, который является корреляционной матрицей для двух переменных:

|             |   age |   lw |
|:------------|------:|-----:|
| **66** **age** |  1    | 0.3  |
| **66** **lw**  |  0.3  | 1    |

Чтобы забрать отсюда коэффициент 0.3, можно через метод `.iloc` выбрать элемент в первой строке и втором столбце или элемент во второй строке и первом столбце:

```python
df.groupby("year")[["age", "lw"]].apply(lambda x: x.corr().iloc[0, 1].round(2))
```

```python
df.groupby("year")[["age", "lw"]].apply(lambda x: x.corr().iloc[1, 0].round(2))
```

Теперь мы получили коэффициенты Пирсона между возрастом и заработной платой за каждый год отдельно.
