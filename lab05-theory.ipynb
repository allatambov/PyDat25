### Простые гистограммы в pandas

Для построения гистограмм в `pandas` используется метод `.hist()`. Он работает как с отдельными столбцами (класс `pandas Series`), так и с датафреймами (класс `pandas DataFrame`). В случае работы с датафреймами в `.hist()` можно выставить основание группировки и построить гистограммы по группам.

Пусть есть датафрейм `data`, в котором зафикисрован возраст и пол респондентов:

```python
import pandas as pd

data = pd.DataFrame([[48, 38, 22, 26, 20, 
                      21, 18, 24, 36, 31, 
                      28, 39, 44, 18, 37], 
                     ["F", "M", "F", "M", "M", 
                      "F", "F", "M", "M", "F",
                      "M", "M", "F", "M", "M"]]).T
data.columns = ["age", "sex"]
data
```

Построим гистограмму для столбца `age`, изменив цвет завливки `color` и цвет границ столбцов (`edgecolor`) и выключив координатную сетку, которая без дополнительных настроек перекрывает столбцы (`grid`):

```python
data["age"].hist(color = "cornflowerblue", 
                 edgecolor = "white", 
                 grid = False);
```

Чтобы построить гистограмму для возраста `age` по группам (одна гистограмма для возраста женщин, одна – для возраста мужчин), метод `.hist()` нужно применить ко всему датафрейму, а в аргументе `by` зафиксировать название столбца, который мы используем как основание группировки:

```python
data.hist("age", by = "sex", 
          color = "cornflowerblue", 
          edgecolor = "white");
```

Гистограммы получились не очень удачными (значения по вертикальной оси не могут быть дробными, это частоты, количество наблюдений, попадающих в тот или иной интервал), но более детальными настройками графиков мы займёмся на занятиях.
