Так как модель из задачи №5 выполняет задачу классификации – присваивает слогам метки «ударный» или «безударный» – для оценки качества ее работы можно использовать следующие метрики:

* точность (*precision*): доля правильно предсказанных ударных слогов относительно всех слогов, которые были отмечены моделью как ударные;
* полнота (*recall*): доля правильно предсказанных ударных слогов относительно всех слогов, которые на самом деле являются ударными.

Чтобы объединить эти метрики в одну, обычно вычисляют F-меру (*F1 score*):

$$
F = 2 \times \frac{\text{precision}\times \text{recall}}{\text{precision}+\text{recall}}.
$$
